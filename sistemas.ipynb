```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir parámetros del sistema
t = np.linspace(0, 1, 1000)  # Tiempo
A = 1.5                      # Amplitud
f = 5.0                      # Frecuencia (Hz)
ruido = 0.4                  # Nivel de ruido

# 2. Construir la señal física y ruido
nt = ruido * np.random.randn(len(t))
xt = A * np.sin(2 * np.pi * f * t) + nt

# 3. Graficar señal resultante
plt.figure(figsize=(6, 3))
plt.plot(t, xt, color='#D47A3A', label='Señal')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#codigo del dia 10-09-2026
import numpy as np
import matplotlib.pyplot as plt

# Configuración del circuito RC
R = 10000          # 10 kohms
C = 50e-6          # 50 microFaradios
tau = R * C        # tau = 0.5 segundos
Vi = 5.0           # Voltaje de entrada

# Vector de tiempo (de 0 a 5 tau)
t = np.linspace(0, 5 * tau, 1000)

# Ecuación de carga del capacitor: Vc(t) = Vi * (1 - e^(-t / tau))
Vc = Vi * (1 - np.exp(-t / tau))

# Graficar respuesta transitoria
plt.figure(figsize=(6, 3))
plt.plot(t, Vc, label='Vc(t)', color='#D47A3A', lw=2)
plt.axvline(tau, color='#5A6B7C', ls='--', label='tau = 0.5s')
plt.xlabel('Tiempo (s)')
plt.ylabel('Voltaje (V)')
plt.title('Carga de Capacitor en Circuito RC')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Voltaje de entrada
Vi = 5.0

# Caso 1: R = 10 kohms, C = 50 microFaradios
R1 = 10000
C1 = 50e-6
tau1 = R1 * C1

# Caso 2: Aumentamos R
R2 = 20000
C2 = 50e-6
tau2 = R2 * C2

# Caso 3: Aumentamos C
R3 = 10000
C3 = 100e-6
tau3 = R3 * C3

# Vector de tiempo
t = np.linspace(0, 5, 1000)

# Ecuaciones de carga
Vc1 = Vi * (1 - np.exp(-t / tau1))
Vc2 = Vi * (1 - np.exp(-t / tau2))
Vc3 = Vi * (1 - np.exp(-t / tau3))

# Graficar
plt.figure(figsize=(8, 5))

plt.plot(t, Vc1, label='R=10kΩ, C=50µF')
plt.plot(t, Vc2, label='R=20kΩ, C=50µF')
plt.plot(t, Vc3, label='R=10kΩ, C=100µF')

plt.xlabel('Tiempo (s)')
plt.ylabel('Voltaje (V)')
plt.title('Comparación de la carga del capacitor')
plt.legend()
plt.grid(True)

plt.show()
```
