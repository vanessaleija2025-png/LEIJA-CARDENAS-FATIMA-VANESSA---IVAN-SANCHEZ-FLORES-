```python
#codigo 5
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Datos de entrada XOR sintéticos
X = np.array([[0,0], [0,1], [1,0], [1,1]], dtype=np.float32)
y = np.array([[0], [1], [1], [0]], dtype=np.float32)

# 1. Construcción del modelo
model = Sequential()
# Capa Dense con 4 neuronas, input_shape=(2,) y activación 'relu'
model.add(Dense(4, input_shape=(2,), activation='relu'))

# Capa de salida Dense con 1 neurona y activación 'sigmoid'
model.add(Dense(1, activation='sigmoid'))

# 2. Compilación del modelo
# Compilación usando optimizador 'adam', pérdida 'binary_crossentropy' y métrica ['accuracy']
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 3. Entrenamiento
model.fit(X, y, epochs=500, verbose=0)

print("Evaluación final del modelo en XOR:")
print(model.predict(X).round())

# PREGUNTA 1: ¿Por qué la función de activación 'sigmoid' es adecuada para la última capa en clasificación binaria?
# RESPUESTA 1: Porque la función sigmoide acota cualquier valor de entrada al rango (0, 1), lo que permite interpretar su salida directamente como una probabilidad de que la muestra pertenezca a la clase positiva (clase 1).
```

```python
#codigo 6
import torch
import torch.nn as nn

class MiRedNeuronal(nn.Module):
    def __init__(self):
        super(MiRedNeuronal, self).__init__()
        # Capa lineal de 10 entradas a 5 salidas ocultas
        self.oculta = nn.Linear(10, 5)

        # Capa de salida lineal de 5 entradas a 1 salida
        self.salida = nn.Linear(5, 1)

        self.relu = nn.ReLU()

    def forward(self, x):
        # Aplica la capa oculta, luego la activación ReLU y finalmente la salida
        x = self.oculta(x)
        x = self.relu(x)
        x = self.salida(x)
        return x

# Instanciación
net = MiRedNeuronal()
print("Estructura de la red en PyTorch:\n", net)

# PREGUNTA 1: ¿Cuál es la función del método 'forward' en la arquitectura de PyTorch?
# RESPUESTA 1: Define el flujo de los datos y cómo se realizan las operaciones de propagación hacia adelante (forward pass) de las entradas a través de las diferentes capas para calcular la salida de la red.
```
