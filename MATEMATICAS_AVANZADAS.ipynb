```python
# ==========================================
# AJUSTANDO UN MODELO DE DESCENSO POR GRADIENTE
# ==========================================

import matplotlib.pyplot as plt

# Función: f(x) = x²
def funcion(x):
    return x**2


# Algoritmo de descenso por gradiente
def gradiente_descendiente(x, learning_rate, iteraciones):

    historia = []

    for i in range(iteraciones):

        # Gradiente de f(x) = x²
        gradiente = 2 * x

        # Guardar el valor actual
        historia.append(x)

        # Actualizar x
        x = x - learning_rate * gradiente

    return x, historia


# ==========================================
# DATOS DEL EJERCICIO
# ==========================================

# Valor inicial
x_inicial = 10

# Tasas de aprendizaje
tasas_aprendizaje = [0.001, 0.01, 0.1, 0.5, 1.0]

# Número de iteraciones
iteraciones = 20


# ==========================================
# RESULTADOS
# ==========================================

for tasa in tasas_aprendizaje:

    resultado, historia = gradiente_descendiente(
        x_inicial,
        tasa,
        iteraciones
    )

    print("Tasa de aprendizaje:", tasa)
    print("Valor inicial:", x_inicial)
    print("Valor final:", resultado)
    print("----------------------------------------")


# ==========================================
# GRÁFICA
# ==========================================

plt.figure(figsize=(10, 6))

for tasa in tasas_aprendizaje:

    resultado, historia = gradiente_descendiente(
        x_inicial,
        tasa,
        iteraciones
    )

    plt.plot(
        range(iteraciones),
        historia,
        marker='o',
        label=f"Tasa = {tasa}"
    )


plt.axhline(0, linestyle='--')

plt.title("Descenso por Gradiente")
plt.xlabel("Iteraciones")
plt.ylabel("Valor de X")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#codigo 2
# Descenso por gradiente
# Función: f(x) = x^2 - 6x + 13

# Definimos la función
def f(x):
    return x**2 - 6*x + 13

# Definimos la derivada
def gradiente(x):
    return 2*x - 6

# Parámetros
x = 10              # Valor inicial
alpha = 0.1         # Tasa de aprendizaje
iteraciones = 50    # Número de iteraciones

# Descenso por gradiente
for i in range(iteraciones):
    x = x - alpha * gradiente(x)

    print(f"Iteración {i+1}: x = {x:.6f}, f(x) = {f(x):.6f}")

# Resultado final
print("\nResultado final:")
print(f"x mínimo = {x:.6f}")
print(f"Valor mínimo de f(x) = {f(x):.6f}")
```

```python
#codigo del dia 10-09-2026
# Implementación en Python - scipy.optimize
import numpy as np
from scipy.optimize import linprog

# Coeficientes de la func. objetivo (-Z)
c = [-3, -2]  # para maximizar Z = 3x1 + 2x2

# Matriz de restricciones de desigualdad (A_ub)
A = [
    [1, 1],  # x1 + x2 <= 4
    [2, 1]   # 2x1 + x2 <= 5
]

# Vector de limites superiores (b_ub)
b = [4, 5]

# Resolver problema lineal (X1>=0, X2>=0)
resultado = linprog(
    c,
    A_ub=A, b_ub=b,
    bounds=(0, None),  # así como en la foto, aplica a las dos variables
    method='highs'
)

print('Solucion Optima:', resultado.x) # te da [1. 3.]
print('Beneficio Maximo Z:', -resultado.fun) # IMPORTANTE: con - porque fun viene negativo
```
