```python
# Análisis de Estabilidad - Sistema Estable
import numpy as np
from scipy.integrate import solve_ivp

# Matriz A con Re(lambda) < 0
A = np.array([[-2.0,  1.0],
              [-1.0, -3.0]])

# Calcular autovalores
eigvals, _ = np.linalg.eig(A)
print('Autovalores:', eigvals)
# Out: [-1.618, -3.382]  (Ambos < 0 -> ESTABLE)

# Simular respuesta x(t)
def system(t, x): return A @ x
sol = solve_ivp(system, [0, 5], [2.0, -1.0])
print('Estado Final x(5):', sol.y[:, -1])
```

```python
# Análisis de Estabilidad - Sistema Inestable
import numpy as np
from scipy.integrate import solve_ivp

# Matriz A con un autovalor Re(lambda) > 0
A_in = np.array([[ 1.5,  2.0],
                 [-0.5, -1.0]])

eigvals_in, _ = np.linalg.eig(A_in)
print('Autovalores:', eigvals_in)
# Out: [ 1.0, -0.5 ]  (1.0 > 0 -> INESTABLE)

# Simular respuesta x(t)
def sys_un(t, x): return A_in @ x
sol_un = solve_ivp(sys_un, [0, 5], [0.1, 0.1])
print('Estado Final x(5):', sol_un.y[:, -1])
```

```python
# Taller 2: Sistema de Segundo Orden
# x_ddot + 3*x_dot + 9*x = 9*u
# H(s) = 9 / (s^2 + 3*s + 9)
import numpy as np
import scipy.signal as signal

num = [9]
den = [1, 3, 9]

wn = np.sqrt(den[2])                     # wn = 3.0 rad/s
zeta = den[1] / (2 * wn)                 # zeta = 0.5 (Subamortiguado)

poles = np.roots(den)                    # p = -1.5 +/- j 2.598
Mp = np.exp(-np.pi * zeta / np.sqrt(1 - zeta**2)) * 100

print(f"wn: {wn:.2f} rad/s, zeta: {zeta:.2f}")
print(f"Polos: {poles}")
print(f"Sobrepaso maximo Mp: {Mp:.2f}%")
```

```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1.0 - s)

def neuron(x, w, b, activation='sigmoid'):
    z = np.dot(w, x) + b
    if activation == 'sigmoid':
        y = sigmoid(z)
        sens = sigmoid_derivative(z)
    return z, y, sens

# Ejemplo de evaluación
x = np.array([1.5, -0.5])
w = np.array([0.8, 1.2])
b = -0.2

z, y, sens = neuron(x, w, b)
print(f"z={z:.3f}, y={y:.3f}, df/dz={sens:.3f}")
```

```python
import numpy as np

np.random.seed(42)
n_in, n_hidden, n_out = 2, 3, 1

# Inicialización Xavier
W1 = np.random.randn(n_in, n_hidden) * np.sqrt(2.0 / n_in)
b1 = np.zeros((1, n_hidden))
W2 = np.random.randn(n_hidden, n_out) * np.sqrt(2.0 / n_hidden)
b2 = np.zeros((1, n_out))

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

# Forward Propagation
X = np.array([[0.5, -0.2]])
z1 = np.dot(X, W1) + b1
a1 = sigmoid(z1)
z2 = np.dot(a1, W2) + b2
y_hat = sigmoid(z2)

print("Activación Oculta (a1):", a1)
print("Predicción Final (y_hat):", y_hat)
```

```python
def sigmoid_deriv(a):
    return a * (1.0 - a)

y = np.array([[1.0]])
learning_rate = 0.5

# 1. Error y Gradiente en Capa de Salida
error = y_hat - y
delta2 = error * sigmoid_deriv(y_hat)
dW2 = np.dot(a1.T, delta2)
db2 = np.sum(delta2, axis=0, keepdims=True)

# 2. Propagación de Error a Capa Oculta
delta1 = np.dot(delta2, W2.T) * sigmoid_deriv(a1)
dW1 = np.dot(X.T, delta1)
db1 = np.sum(delta1, axis=0, keepdims=True)

# 3. Actualización de Pesos
W2 -= learning_rate * dW2
b2 -= learning_rate * db2
W1 -= learning_rate * dW1
b1 -= learning_rate * db1

print("Gradiente dW2:\n", dW2)
print("Gradiente dW1:\n", dW1)
```

Este código simula el comportamiento de una sola neurona artificial (el componente más básico de una red neuronal).

La función Sigmoide (sigmoid): Es una función matemática que toma cualquier número (positivo o negativo) y lo transforma en un valor entre 0 y 1. Esto es muy útil porque permite interpretar el resultado final como una probabilidad (por ejemplo, un 80% de probabilidad de que algo sea correcto).
La derivada de la Sigmoide (sigmoid_derivative): Sirve para medir qué tanto cambia la salida de la neurona si hacemos un pequeño ajuste en la entrada. Es como medir la sensibilidad de la neurona.
La función Neurona (neuron): Junta todo el proceso.
Primero, multiplica cada dato de entrada por su respectivo nivel de importancia (los pesos) y le suma un valor de ajuste (el sesgo).
Segundo, pasa ese resultado por la función sigmoide para obtener la activación final (la predicción).
Tercero, calcula la sensibilidad de la neurona para saber cómo ajustar las cosas en el futuro.
Al final, con las entradas y pesos de prueba que pusiste, la neurona procesa la información y nos dice que el resultado final es aproximadamente 0.599 (cerca del 60%) con una sensibilidad de 0.240.

ANALISIS DEL SEGUNDO CODIGO: RED NEURONAL MULTICAPA Y ENTRENAMIENTO (Celdas 2sD3hyff3nsg y Fzk3NH-v3wpP)
Aquí ya no tenemos una sola neurona, sino una red pequeña completa con tres capas: una de entrada (recibe los datos), una oculta (procesa y busca patrones) y una de salida (da la respuesta final). El proceso se divide en dos grandes pasos:

Paso 1: Ir hacia adelante (Forward Propagation - Celda 2sD3hyff3nsg)
Inicialización: Se configuran los pesos iniciales de la red usando un método llamado Xavier. Esto asegura que la red empiece con valores aleatorios pero equilibrados, evitando que la información se pierda o se vuelva gigante desde el inicio.
Cálculo de la capa oculta: Los datos de entrada pasan a las neuronas de la capa intermedia (oculta). Aquí se multiplican por los primeros pesos y se les aplica la función sigmoide.
Cálculo de la salida: Lo que procesó la capa oculta se envía a la neurona final de salida para darnos la predicción inicial de la red, que en este ejemplo da como resultado 0.670.
Paso 2: Aprender de los errores e ir hacia atrás (Backpropagation - Celda Fzk3NH-v3wpP)
Una vez que la red da su predicción (0.670), se da cuenta de que el valor real que debía dar era 1.0. Es decir, cometió un error.

Calcular el Error: Restamos la predicción obtenida menos el valor real. La red analiza qué tan lejos se quedó de la meta.
Propagar el error hacia atrás: Utilizando la derivada (la sensibilidad que explicamos en el primer código), la red calcula cuánta culpa tuvo cada peso (conexión) en haber cometido ese error. Esto se calcula primero para la capa de salida y luego se distribuye hacia atrás a la capa oculta.
Actualizar los Pesos: Usando una tasa de aprendizaje (que es la velocidad a la que aprende la red, definida aquí en 0.5), el código ajusta ligeramente todos los pesos y sesgos restándoles una parte del error.
